$flow[\frac{m}{yr}]=\frac{\Delta E}{\Delta x}\times C$

$\Delta x$ is slope

Break up ice sheet into 'cells'

Each cell has an elevation. Between cells, ice flows into and out of them from their neighbours. The edges have 0 elevation.

When there is a difference in elevation between two cells, ice flows between them

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
```

```python
nX = 10                # number of grid points
domainWidth = 1e6      # meters
timeStep = 100         # years
nYears = 50000         # years
flowParam = 1e4        # m horizontal / yr
snowFall = 0.5         # m / y
dX = domainWidth / nX  # m / grid point
```

```python
elevations = np.zeros(nX+2)
flows = np.zeros(nX+1)
```

```python
def calcFlows():
    for i in range (nX + 2 - 1):
        flows[i] = ( elevations[i] - elevations[i+1] ) / dX * flowParam  * \
            ( elevations[i]+elevations[i+1] ) / 2 / dX
```

```python
def calcElevations():
    for i in range (1, nX + 2 - 1):
        elevations[i] += ( snowFall + flows[i-1] - flows[i] ) * timeStep
```

```python
def simulateFrame():
    calcFlows()
    calcElevations()
```

```python
frameDuration = 0.001    # s
duration = nYears / timeStep * frameDuration
plotLimit = int(snowFall * nYears / 2)
interval = 1
total_frames = int(duration / frameDuration / interval)
def plot():
    # Reset the elevations
    elevations[:] = 0

    fig, ax = plt.subplots()
    line, = ax.plot(elevations, 'b-o', lw=2)
    ax.set_ylim([0,plotLimit])

    ax.set_title("Ice Sheet Profile Evolution")
    ax.set_xlabel("Grid Cell")
    ax.set_ylabel("Elevation (m)")

    def animateFrame(frame):
        simulateFrame()
        line.set_ydata(elevations)
        return line,

    ani = FuncAnimation(fig, animateFrame, frames=total_frames, interval=interval, blit=True)
    plt.close(fig)

    display(HTML(ani.to_jshtml()))
```

```python
plot()
print(elevations[int(len(elevations) / 2)])
```
